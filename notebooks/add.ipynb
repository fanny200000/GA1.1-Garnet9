```python
def sum_list(values):
    total = 0
    for value in values:
        total = add(total, value) 
    return total
```

```python
values = [1,2,3,4,5]
print(sum_list(values))

values = [-1,-2,-3,-4,-5]
print(sum_list(values))
```
